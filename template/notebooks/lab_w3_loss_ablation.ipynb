## 1. Setup

Semua import, model, loss, dataset toy, dan training loop dikumpulkan di satu sel. Sel berikutnya hanya memakai objek yang sudah disiapkan di sini.


```python
import itertools
from pathlib import Path

import yaml
import torch
import torch.nn as nn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader, TensorDataset

# Output ditulis relatif ke working directory Colab/notebook, bukan ke folder repo.
OUTPUT_DIR = Path("lab3_outputs").resolve()
OUTPUT_DIR.mkdir(exist_ok=True)


class FocalLoss(torch.nn.Module):
    """Multi-class focal loss; gamma=0 dan alpha=None identik dengan CrossEntropyLoss."""

    def __init__(self, gamma=2.0, alpha=None, label_smoothing=0.0):
        super().__init__()
        self.gamma = gamma
        self.label_smoothing = label_smoothing
        if alpha is not None:
            self.register_buffer("alpha", torch.tensor(alpha, dtype=torch.float32))
        else:
            self.alpha = None

    def forward(self, logits, targets):
        ce = torch.nn.functional.cross_entropy(
            logits, targets, reduction="none", label_smoothing=self.label_smoothing
        )
        if self.gamma == 0.0 and self.alpha is None:
            return ce.mean()
        with torch.no_grad():
            probs = torch.nn.functional.softmax(logits, dim=1)
            pt = probs.gather(1, targets.unsqueeze(1)).squeeze(1).clamp_min(1e-8)
        loss = ((1.0 - pt) ** self.gamma) * ce
        if self.alpha is not None:
            loss = self.alpha.to(loss.device)[targets] * loss
        return loss.mean()


class SimpleCNN(nn.Module):
    def __init__(self, num_classes=10, in_channels=3):
        super().__init__()
        self.block1 = nn.Sequential(
            nn.Conv2d(in_channels, 32, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 64, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
        )
        self.block2 = nn.Sequential(
            nn.Conv2d(64, 128, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 128, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
        )
        self.classifier = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Dropout(0.2),
            nn.Linear(128, num_classes),
        )

    def forward(self, x):
        x = self.block1(x)
        x = self.block2(x)
        return self.classifier(x)


def make_toy_images(seed=0, n_train=640, n_val=240, num_classes=10):
    """Dataset toy yang learnable: tiap kelas punya patch terang di lokasi berbeda.

    Label tidak acak murni. CNN bisa belajar pola spasial sederhana, sehingga hasil
    ablation lebih masuk akal daripada random-label toy data.
    """
    g = torch.Generator().manual_seed(seed)
    total = n_train + n_val
    labels = torch.arange(total) % num_classes
    labels = labels[torch.randperm(total, generator=g)]

    images = 0.20 * torch.randn(total, 3, 32, 32, generator=g)
    grid_positions = [(r, c) for r in [3, 11, 19, 25] for c in [3, 11, 19]]
    for i, label in enumerate(labels):
        y0, x0 = grid_positions[int(label) % len(grid_positions)]
        images[i, :, y0:y0 + 5, x0:x0 + 5] += 1.5
        images[i, int(label) % 3, :, :] += 0.05
    images = images.clamp(-1.0, 2.0)

    x_train, y_train = images[:n_train], labels[:n_train]
    x_val, y_val = images[n_train:], labels[n_train:]
    return (
        DataLoader(TensorDataset(x_train, y_train), batch_size=64, shuffle=True),
        DataLoader(TensorDataset(x_val, y_val), batch_size=64),
    )


def train_condition(loss_name, freeze_until, seed, epochs=3):
    torch.manual_seed(seed)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    train_loader, val_loader = make_toy_images(seed=seed)
    model = SimpleCNN(num_classes=10).to(device)
    if freeze_until == "block1":
        for p in model.block1.parameters():
            p.requires_grad = False
    loss_fn = FocalLoss(gamma=2.0) if loss_name == "focal" else nn.CrossEntropyLoss()
    opt = torch.optim.AdamW([p for p in model.parameters() if p.requires_grad], lr=1e-3, weight_decay=1e-4)
    for _ in range(epochs):
        model.train()
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            opt.zero_grad(set_to_none=True)
            loss = loss_fn(model(x), y)
            loss.backward()
            opt.step()
    model.eval()
    correct = total = 0
    with torch.no_grad():
        for x, y in val_loader:
            x, y = x.to(device), y.to(device)
            pred = model(x).argmax(1)
            correct += (pred == y).sum().item()
            total += y.numel()
    return correct / max(total, 1)


print("Output folder:", OUTPUT_DIR)
print("CUDA tersedia:", torch.cuda.is_available())
```

## 2. Mulai dari satu tensor: `FocalLoss(gamma=0)` == `CrossEntropyLoss`

Sebelum menjalankan 12 run, uji dulu blok pembangun paling kecil. Focal loss dengan `gamma=0` harus sama dengan cross-entropy. Jika bagian ini gagal, eksperimen belum layak dijalankan.


```python
torch.manual_seed(0)

logits = torch.randn(16, 10)
labels = torch.randint(0, 10, (16,))

ce_loss = torch.nn.CrossEntropyLoss()(logits, labels).item()
focal_loss = FocalLoss(gamma=0.0)(logits, labels).item()
diff = abs(ce_loss - focal_loss)

print(f"CrossEntropy : {ce_loss:.6f}")
print(f"Focal gamma=0: {focal_loss:.6f}")
print(f"Selisih      : {diff:.6f}")

assert diff < 1e-4, f"FocalLoss(gamma=0) tidak sama dengan CE. Selisih: {diff}"
print("OK: sanity check lulus.")
```

## 3. Definisikan satu baseline dict

Baseline ditulis sebagai dictionary di notebook, bukan dibaca dari file YAML.


```python
base_cfg = {
    "experiment_name": "lab3_baseline",
    "model": {"name": "simple_cnn", "num_classes": 10, "freeze_until": None},
    "loss": {"name": "cross_entropy", "label_smoothing": 0.0},
    "train": {"epochs": 3},
}

print("Experiment name:", base_cfg["experiment_name"])
print("Model          :", base_cfg["model"]["name"])
print("Loss baseline  :", base_cfg["loss"]["name"])
print("Epoch baseline :", base_cfg["train"]["epochs"])
```

## 4. Ubah satu faktor: loss

Pertama ubah hanya loss. Ini membuat hubungan CE dan focal terlihat jelas sebelum faktor freeze ditambahkan.

```python
def clone_cfg(cfg):
    return yaml.safe_load(yaml.dump(cfg))


def make_cfg(base_cfg, loss_name, freeze_until):
    cfg = clone_cfg(base_cfg)
    freeze_label = freeze_until or "none"
    exp_name = f"lab3_{loss_name}_freeze{freeze_label}"

    cfg["experiment_name"] = exp_name
    cfg["loss"] = {"name": loss_name}
    if loss_name == "focal":
        cfg["loss"]["gamma"] = 2.0
    cfg["model"]["freeze_until"] = freeze_until

    return exp_name, freeze_label, cfg


for loss_name in ["cross_entropy", "focal"]:
    exp_name, freeze_label, cfg = make_cfg(base_cfg, loss_name, None)
    print(exp_name, "->", cfg["loss"])
```

## 5. Tambah faktor kedua: freeze

Setelah loss jelas, tambahkan faktor freeze. Kombinasi 2 loss x 2 freeze menghasilkan 4 kondisi config.

```python
losses = ["cross_entropy", "focal"]
freezes = [None, "block1"]

conditions = []
for loss_name, freeze_until in itertools.product(losses, freezes):
    exp_name, freeze_label, cfg = make_cfg(base_cfg, loss_name, freeze_until)
    conditions.append(
        {
            "loss": loss_name,
            "freeze": freeze_label,
            "experiment_name": exp_name,
            "config": cfg,
        }
    )

print(f"Jumlah kondisi: {len(conditions)}")
for c in conditions:
    print(f"{c['experiment_name']:36s} loss={c['loss']:15s} freeze={c['freeze']}")
```

## 6. Susun 12 run

Setiap kondisi diulang dengan 3 seed. Pada tahap ini kita baru membuat daftar run. Training belum dijalankan.


```python
seeds = [42, 123, 2024]
runs = []
for c in conditions:
    for seed in seeds:
        runs.append(
            {
                "loss": c["loss"],
                "freeze": c["freeze"],
                "seed": seed,
                "experiment_name": c["experiment_name"],
            }
        )

print(f"Kondisi: {len(conditions)}")
print(f"Seed per kondisi: {len(seeds)}")
print(f"Total run: {len(runs)}")
for r in runs:
    print(f"{r['experiment_name']} | seed={r['seed']}")
```

## 7. Jalankan toy ablation

Sel berikut menjalankan eksperimen langsung dari notebook. Dataset yang dipakai adalah toy image dataset kecil.

Default `RUN_NOW=False` supaya notebook tidak langsung menjalankan semua run ketika dibuka. Ubah menjadi `True` setelah rencana eksperimen diisi.


```python
RUN_NOW = False

if RUN_NOW:
    rows = []
    for r in runs:
        print(f"Running: {r['experiment_name']} seed={r['seed']}")
        val_acc = train_condition(r["loss"], None if r["freeze"] == "none" else r["freeze"], r["seed"])
        rows.append({**r, "best_val_acc": val_acc})
        print(f"  val_acc={val_acc:.4f}")

    results_path = OUTPUT_DIR / "lab3_results.csv"
    pd.DataFrame(rows).to_csv(results_path, index=False)
    print(f"Hasil ditulis ke: {results_path}")
else:
    print("RUN_NOW=False. Isi rencana eksperimen, lalu ubah menjadi True untuk menjalankan 12 toy run.")
```

## 8. Rangkum hasil

Setelah 12 run selesai, `lab3_outputs/lab3_results.csv` berisi satu baris per run. Sel ini memfilter run Lab 3, mengurai nama eksperimen menjadi kolom `loss` dan `freeze`, lalu menghitung mean +/- std dari 3 seed.


```python
results_path = OUTPUT_DIR / "lab3_results.csv"

if results_path.exists():
    df = pd.read_csv(results_path)
    df_lab3 = df[df["experiment_name"].str.startswith("lab3_")].copy()

    def parse_exp(name):
        parts = name.replace("lab3_", "").rsplit("_freeze", 1)
        return parts[0], parts[1] if len(parts) > 1 else "none"

    df_lab3[["loss", "freeze"]] = df_lab3["experiment_name"].apply(
        lambda name: pd.Series(parse_exp(name))
    )
    df_lab3["best_val_acc"] = df_lab3["best_val_acc"].astype(float)

    summary = df_lab3.groupby(["loss", "freeze"])["best_val_acc"].agg(["mean", "std", "count"])
    summary.columns = ["mean_val_acc", "std_val_acc", "n_seeds"]
    summary = summary.reset_index()
    summary["mean_pct"] = summary["mean_val_acc"] * 100
    summary["std_pct"] = summary["std_val_acc"].fillna(0.0) * 100
else:
    summary = pd.DataFrame(columns=["loss", "freeze", "mean_val_acc", "std_val_acc", "n_seeds", "mean_pct", "std_pct"])
    print("Belum ada lab3_results.csv. Jalankan sel eksperimen dengan RUN_NOW=True untuk membuat hasil.")

if summary.empty:
    print("Tabel hasil Lab 3 masih kosong.")
else:
    print("Tabel hasil Lab 3:")
    print(summary.to_string(index=False))
```

## 9. Bar chart dengan error bar

```python
if summary.empty:
    print("Plot dilewati karena summary masih kosong. Jalankan RUN_NOW=True terlebih dahulu.")
else:
    fig, ax = plt.subplots(figsize=(8, 5))

    conditions_label = [f"{r['loss']}\nfreeze={r['freeze']}" for _, r in summary.iterrows()]
    means = summary["mean_pct"].values
    stds = summary["std_pct"].values

    colors = ["#4C8BB5", "#E07B54", "#5BAD6F", "#C25B99"]
    bars = ax.bar(
        conditions_label,
        means,
        yerr=stds,
        capsize=6,
        color=colors[: len(conditions_label)],
        alpha=0.85,
        edgecolor="white",
        linewidth=1.5,
    )

    baseline_row = summary[
        (summary["loss"] == "cross_entropy") & (summary["freeze"] == "none")
    ]
    if not baseline_row.empty:
        baseline_mean = baseline_row["mean_pct"].values[0]
        ax.axhline(
            baseline_mean,
            color="gray",
            linestyle="--",
            alpha=0.6,
            label=f"Baseline: {baseline_mean:.1f}%",
        )

    for bar, mean, std in zip(bars, means, stds):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + std + 0.3,
            f"{mean:.1f}+/-{std:.1f}",
            ha="center",
            va="bottom",
            fontsize=9,
        )

    ymin = max(0, min(means) - max(stds) - 3)
    ymax = min(100, max(means) + max(stds) + 5)
    ax.set_ylim(ymin, ymax)
    ax.set_ylabel("Val Accuracy (%) - mean +/- std dari 3 seed")
    ax.set_title("Lab 3: Toy Ablation Loss x Freeze")
    ax.legend()
    ax.grid(axis="y", alpha=0.3)

    plot_dir = OUTPUT_DIR / "plots"
    plot_dir.mkdir(parents=True, exist_ok=True)
    plt.tight_layout()
    plt.savefig(plot_dir / "lab3_ablation.png", dpi=120, bbox_inches="tight")
    plt.show()
```

## 10. Analisis main effect dan interaksi

Main effect menjawab: jika satu faktor dirata-ratakan atas faktor lain, apakah faktor itu tetap terlihat berpengaruh?

Interaksi menjawab: apakah pengaruh focal loss berubah ketika freeze dinyalakan?

Karena dataset ini toy, tafsirkan hasil sebagai latihan membaca pola eksperimen, bukan sebagai klaim final tentang model pada dataset nyata.


```python
if summary.empty:
    print("Analisis efek dilewati karena summary masih kosong.")
else:
    effect_loss = summary.groupby("loss")["mean_pct"].mean()
    print("Main effect loss:")
    print(effect_loss.to_string())
    print(f"Delta focal - CE: {effect_loss.get('focal', 0) - effect_loss.get('cross_entropy', 0):+.2f}%")

    print()

    effect_freeze = summary.groupby("freeze")["mean_pct"].mean()
    print("Main effect freeze:")
    print(effect_freeze.to_string())
    print(f"Delta freeze block1 - none: {effect_freeze.get('block1', 0) - effect_freeze.get('none', 0):+.2f}%")

    print()

    pivot = summary.pivot(index="freeze", columns="loss", values="mean_pct")
    print("Pivot table (mean val_acc %):")
    print(pivot.round(2))

    if {"focal", "cross_entropy"}.issubset(pivot.columns) and {"none", "block1"}.issubset(pivot.index):
        diff_no_freeze = pivot.loc["none", "focal"] - pivot.loc["none", "cross_entropy"]
        diff_freeze = pivot.loc["block1", "focal"] - pivot.loc["block1", "cross_entropy"]
        interaction = diff_freeze - diff_no_freeze

        print(f"\nEfek focal tanpa freeze : {diff_no_freeze:+.2f}%")
        print(f"Efek focal dengan freeze: {diff_freeze:+.2f}%")
        print(f"Interaksi              : {interaction:+.2f}%")

        if abs(interaction) > 1.0:
            print("Interaksi cukup besar: efek loss bergantung pada kondisi freeze.")
        else:
            print("Interaksi kecil: efek loss relatif mirip pada kedua kondisi freeze.")
```

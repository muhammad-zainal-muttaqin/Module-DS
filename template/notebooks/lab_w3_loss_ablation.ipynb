## 1. Setup

Semua import dan penyetelan path dikumpulkan di satu sel. Sel berikutnya hanya memakai objek yang sudah disiapkan di sini.

````python
import os
import sys
import subprocess
import itertools
from pathlib import Path

import yaml
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Notebook ini diasumsikan dijalankan dari folder template/notebooks.
ROOT = Path("..").resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.losses import FocalLoss

print("Root template:", ROOT)
print("CUDA tersedia:", torch.cuda.is_available())
````

## 2. Mulai dari satu tensor: `FocalLoss(gamma=0)` ≡ `CrossEntropyLoss`

Sebelum menjalankan 12 run, uji dulu blok pembangun paling kecil. Focal loss dengan `gamma=0` harus sama dengan cross-entropy. Jika bagian ini gagal, eksperimen penuh belum layak dijalankan.

````python
torch.manual_seed(0)

logits = torch.randn(16, 10)
labels = torch.randint(0, 10, (16,))

ce_loss = torch.nn.CrossEntropyLoss()(logits, labels).item()
focal_loss = FocalLoss(gamma=0.0)(logits, labels).item()
diff = abs(ce_loss - focal_loss)

print(f"CrossEntropy : {ce_loss:.6f}")
print(f"Focal gamma=0: {focal_loss:.6f}")
print(f"Selisih      : {diff:.6f}")

assert diff < 1e-4, f"FocalLoss(gamma=0) tidak sama dengan CE. Selisih: {diff}"
print("OK: sanity check lulus.")
````

## 3. Baca satu config baseline

Mulai dari `configs/baseline.yaml`. Kita tidak menulis config dari nol karena baseline sudah memuat pilihan data, model, optimizer, scheduler, dan output yang dipakai pada W2.

````python
baseline_path = ROOT / "configs" / "baseline.yaml"

with baseline_path.open(encoding="utf-8") as f:
    base_cfg = yaml.safe_load(f)

print("Experiment name:", base_cfg["experiment_name"])
print("Model          :", base_cfg["model"]["name"])
print("Loss baseline  :", base_cfg["loss"]["name"])
print("Epoch baseline :", base_cfg["train"]["epochs"])
````

## 4. Ubah satu faktor: loss

Pertama ubah hanya loss. Ini membuat hubungan CE dan focal terlihat jelas sebelum faktor freeze ditambahkan.

````python
def clone_cfg(cfg):
    return yaml.safe_load(yaml.dump(cfg))


def make_cfg(base_cfg, loss_name, freeze_until):
    cfg = clone_cfg(base_cfg)
    freeze_label = freeze_until or "none"
    exp_name = f"lab2_{loss_name}_freeze{freeze_label}"

    cfg["experiment_name"] = exp_name
    cfg["train"]["epochs"] = 15
    cfg["loss"] = {"name": loss_name}
    if loss_name == "focal":
        cfg["loss"]["gamma"] = 2.0
    cfg["model"]["freeze_until"] = freeze_until

    return exp_name, freeze_label, cfg


for loss_name in ["cross_entropy", "focal"]:
    exp_name, freeze_label, cfg = make_cfg(base_cfg, loss_name, None)
    print(exp_name, "->", cfg["loss"])
````

## 5. Tambah faktor kedua: freeze

Setelah loss jelas, tambahkan faktor freeze. Kombinasi 2 loss × 2 freeze menghasilkan 4 kondisi config.

````python
losses = ["cross_entropy", "focal"]
freezes = [None, "block1"]

conditions = []
for loss_name, freeze_until in itertools.product(losses, freezes):
    exp_name, freeze_label, cfg = make_cfg(base_cfg, loss_name, freeze_until)
    conditions.append(
        {
            "loss": loss_name,
            "freeze": freeze_label,
            "experiment_name": exp_name,
            "config": cfg,
        }
    )

print(f"Jumlah kondisi: {len(conditions)}")
for c in conditions:
    print(f"{c['experiment_name']:36s} loss={c['loss']:15s} freeze={c['freeze']}")
````

## 6. Tulis 4 config dan susun 12 run

Setiap kondisi diulang dengan 3 seed. Pada tahap ini kita baru membuat file config dan daftar perintah run. Training belum dijalankan.

````python
seeds = [42, 123, 2024]
config_dir = ROOT / "configs" / "lab2"
config_dir.mkdir(exist_ok=True)

runs = []
for c in conditions:
    config_path = config_dir / f"{c['experiment_name']}.yaml"
    with config_path.open("w", encoding="utf-8") as f:
        yaml.safe_dump(c["config"], f, sort_keys=False, allow_unicode=True)

    for seed in seeds:
        runs.append(
            {
                "loss": c["loss"],
                "freeze": c["freeze"],
                "seed": seed,
                "config_path": config_path,
                "experiment_name": c["experiment_name"],
            }
        )

print(f"Kondisi config: {len(conditions)}")
print(f"Seed per kondisi: {len(seeds)}")
print(f"Total run: {len(runs)}")
print()
for r in runs:
    rel_config = r["config_path"].relative_to(ROOT)
    print(f"python -m src.train --config {rel_config.as_posix()} --seed {r['seed']}")
````

## 7. Jalankan eksperimen

Cara yang dianjurkan: jalankan dari terminal di folder `template/` agar progress dan log terlihat penuh.


````python
RUN_NOW = False

if RUN_NOW:
    for r in runs:
        rel_config = r["config_path"].relative_to(ROOT)
        cmd = [
            sys.executable,
            "-m",
            "src.train",
            "--config",
            rel_config.as_posix(),
            "--seed",
            str(r["seed"]),
        ]
        print("Running:", " ".join(cmd))
        result = subprocess.run(cmd, cwd=ROOT, capture_output=True, text=True)
        if result.returncode != 0:
            print(result.stdout[-1000:])
            print(result.stderr[-1000:])
            raise RuntimeError(f"Run gagal: {' '.join(cmd)}")
    print("Semua run selesai.")
else:
    print("RUN_NOW=False. Jalankan dari terminal, lalu lanjutkan ke bagian ringkasan.")
````

## 8. Rangkum hasil

Setelah 12 run selesai, `experiments/results.csv` berisi satu baris per run. Sel ini memfilter run Lab 2, mengurai nama eksperimen menjadi kolom `loss` dan `freeze`, lalu menghitung mean ± std dari 3 seed.

````python
results_path = ROOT / "experiments" / "results.csv"
df = pd.read_csv(results_path)

df_lab2 = df[df["experiment_name"].str.startswith("lab2_")].copy()

def parse_exp(name):
    parts = name.replace("lab2_", "").rsplit("_freeze", 1)
    return parts[0], parts[1] if len(parts) > 1 else "none"


df_lab2[["loss", "freeze"]] = df_lab2["experiment_name"].apply(
    lambda name: pd.Series(parse_exp(name))
)
df_lab2["best_val_acc"] = df_lab2["best_val_acc"].astype(float)

summary = df_lab2.groupby(["loss", "freeze"])["best_val_acc"].agg(["mean", "std", "count"])
summary.columns = ["mean_val_acc", "std_val_acc", "n_seeds"]
summary = summary.reset_index()
summary["mean_pct"] = summary["mean_val_acc"] * 100
summary["std_pct"] = summary["std_val_acc"] * 100

print("Tabel hasil Lab 2:")
print(summary.to_string(index=False))
````

## 9. Bar chart dengan error bar

````python
fig, ax = plt.subplots(figsize=(8, 5))

conditions_label = [f"{r['loss']}\nfreeze={r['freeze']}" for _, r in summary.iterrows()]
means = summary["mean_pct"].values
stds = summary["std_pct"].values

colors = ["#4C8BB5", "#E07B54", "#5BAD6F", "#C25B99"]
bars = ax.bar(
    conditions_label,
    means,
    yerr=stds,
    capsize=6,
    color=colors[: len(conditions_label)],
    alpha=0.85,
    edgecolor="white",
    linewidth=1.5,
)

baseline_row = summary[
    (summary["loss"] == "cross_entropy") & (summary["freeze"] == "none")
]
if not baseline_row.empty:
    baseline_mean = baseline_row["mean_pct"].values[0]
    ax.axhline(
        baseline_mean,
        color="gray",
        linestyle="--",
        alpha=0.6,
        label=f"Baseline: {baseline_mean:.1f}%",
    )

for bar, mean, std in zip(bars, means, stds):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + std + 0.3,
        f"{mean:.1f}±{std:.1f}",
        ha="center",
        va="bottom",
        fontsize=9,
    )

ymin = max(0, min(means) - max(stds) - 3)
ymax = max(means) + max(stds) + 3
ax.set_ylim(ymin, ymax)
ax.set_ylabel("Val Accuracy (%) - mean ± std dari 3 seed")
ax.set_title("Lab 2: Ablation Loss × Freeze")
ax.legend()
ax.grid(axis="y", alpha=0.3)

plot_dir = ROOT / "experiments" / "plots"
plot_dir.mkdir(exist_ok=True)
plt.tight_layout()
plt.savefig(plot_dir / "lab2_ablation.png", dpi=120, bbox_inches="tight")
plt.show()
````

## 10. Analisis main effect dan interaksi

Main effect menjawab: jika satu faktor dirata-ratakan atas faktor lain, apakah faktor itu tetap terlihat berpengaruh?

Interaksi menjawab: apakah pengaruh focal loss berubah ketika freeze dinyalakan?

````python
effect_loss = summary.groupby("loss")["mean_pct"].mean()
print("Main effect loss:")
print(effect_loss.to_string())
print(f"Delta focal - CE: {effect_loss.get('focal', 0) - effect_loss.get('cross_entropy', 0):+.2f}%")

print()

effect_freeze = summary.groupby("freeze")["mean_pct"].mean()
print("Main effect freeze:")
print(effect_freeze.to_string())
print(f"Delta freeze block1 - none: {effect_freeze.get('block1', 0) - effect_freeze.get('none', 0):+.2f}%")

print()

pivot = summary.pivot(index="freeze", columns="loss", values="mean_pct")
print("Pivot table (mean val_acc %):")
print(pivot.round(2))

if {"focal", "cross_entropy"}.issubset(pivot.columns) and {"none", "block1"}.issubset(pivot.index):
    diff_no_freeze = pivot.loc["none", "focal"] - pivot.loc["none", "cross_entropy"]
    diff_freeze = pivot.loc["block1", "focal"] - pivot.loc["block1", "cross_entropy"]
    interaction = diff_freeze - diff_no_freeze

    print(f"\nEfek focal tanpa freeze : {diff_no_freeze:+.2f}%")
    print(f"Efek focal dengan freeze: {diff_freeze:+.2f}%")
    print(f"Interaksi              : {interaction:+.2f}%")

    if abs(interaction) > 1.0:
        print("Interaksi cukup besar: efek loss bergantung pada kondisi freeze.")
    else:
        print("Interaksi kecil: efek loss relatif mirip pada kedua kondisi freeze.")
````

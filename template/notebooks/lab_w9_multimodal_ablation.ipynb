# Lab W9: Multimodal Ablation


## Alur Lab

1. **Muat dua view nyata:** ambil dua representasi fitur dari dataset angka mfeat sebagai dua modalitas.
2. **Lihat datasetnya:** tampilkan citra digit dari modalitas Pixel dan periksa 6 fitur Morfologi agar jelas isi tiap modalitas.
3. **Baseline per modalitas:** uji tiap view sendirian untuk tahu batas performanya.
4. **Late fusion:** gabungkan dua modalitas dan ukur kenaikannya.
5. **Ablation enam kondisi:** matikan sebagian input untuk menguji ketergantungan model.
6. **Modality dropout:** latih model agar tidak rapuh terhadap modalitas hilang.

Lab ini mengimplementasikan **late fusion** pada dataset multimodal nyata, menjalankan **ablation per-modalitas** (6 kondisi), mendeteksi **masalah modalitas terabaikan**, lalu menerapkan **modality dropout** sebagai solusi.

- **Dataset:** mfeat (Multiple Features), angka tulisan tangan 0-9 dari NIST. Tiap sampel punya beberapa view fitur nyata. Lab ini memakai dua view sebagai dua modalitas: intensitas pixel rata-rata (240-dim) yang sinyalnya kuat, dan fitur morfologi seperti ukuran serta ketebalan goresan (6-dim) yang sinyalnya lemah.
- **Target:** klasifikasi digit 0-9.
- **Estimasi waktu:** 2-3 jam.


## 1. Setup & Data


```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device: {DEVICE}")
np.random.seed(42)
torch.manual_seed(42)
```

```python
# Dataset mfeat (Multiple Features): angka tulisan tangan 0-9 dari NIST.
# Tiap sampel punya beberapa "view" fitur nyata. Kita pakai dua view sebagai dua modalitas:
#   - Modalitas A "Pixel"    : 240 intensitas pixel rata-rata (sinyal kuat)
#   - Modalitas B "Morfologi": 6 fitur morfologi seperti ukuran dan ketebalan goresan (sinyal lemah)
# Kedua view berasal dari 2000 sampel yang sama dengan urutan baris yang sama, jadi labelnya sejajar.

pixel = fetch_openml("mfeat-pixel", version=1, as_frame=False)
morph = fetch_openml("mfeat-morphological", version=1, as_frame=False)

# Pastikan kedua view sejajar per baris (label harus identik di tiap baris)
assert (pixel.target == morph.target).all(), "Urutan baris dua view tidak sejajar"

Xa_raw = pixel.data.astype("float32")     # (2000, 240) Pixel
Xb_raw = morph.data.astype("float32")     # (2000, 6)   Morfologi
y_raw = LabelEncoder().fit_transform(pixel.target)  # digit -> 0..9

# Standardisasi tiap modalitas (skala fitur dua view sangat berbeda)
Xa = StandardScaler().fit_transform(Xa_raw).astype("float32")
Xb = StandardScaler().fit_transform(Xb_raw).astype("float32")

D_A, D_B = Xa.shape[1], Xb.shape[1]
NUM_CLASSES = 10

# Split train/val (stratified agar tiap digit terwakili seimbang)
idx = np.arange(len(y_raw))
idx_tr, idx_va = train_test_split(idx, test_size=0.3, random_state=42, stratify=y_raw)

Xa_tr, Xa_va = torch.tensor(Xa[idx_tr]), torch.tensor(Xa[idx_va])
Xb_tr, Xb_va = torch.tensor(Xb[idx_tr]), torch.tensor(Xb[idx_va])
y_tr, y_va = torch.tensor(y_raw[idx_tr]), torch.tensor(y_raw[idx_va])

dl_tr = DataLoader(TensorDataset(Xa_tr, Xb_tr, y_tr), 64, shuffle=True)
dl_va = DataLoader(TensorDataset(Xa_va, Xb_va, y_va), 64, shuffle=False)

print(f"Train: Pixel {tuple(Xa_tr.shape)}, Morfologi {tuple(Xb_tr.shape)}")
print(f"Val:   Pixel {tuple(Xa_va.shape)}, Morfologi {tuple(Xb_va.shape)}")
print(f"Kelas: {NUM_CLASSES} digit (0-9)")
```

## 1b. Kenalan dengan dataset mfeat

Sebelum melatih model, kita lihat dulu data yang akan dipakai supaya jelas apa yang sebenarnya diproses.

**mfeat (Multiple Features)** berisi 2000 citra angka tulisan tangan 0-9 yang dipindai dari peta utilitas Belanda, 200 sampel per digit. Keistimewaannya: tiap angka yang sama direkam dalam beberapa "view", yaitu cara berbeda untuk mendeskripsikan objek yang sama. Dua view inilah yang kita pakai sebagai dua modalitas:

- **Modalitas A - Pixel (240 fitur):** rata-rata intensitas pixel pada grid 16x15 (jendela 2x3 dari citra asli). Karena ini nilai pixel, 240 angkanya bisa di-reshape kembali menjadi gambar 16x15 sehingga bentuk digitnya terlihat. Nilainya 0 sampai 6.
- **Modalitas B - Morfologi (6 fitur):** 6 angka yang merangkum bentuk goresan, seperti ukuran, ketebalan, dan keliling. Hanya 6 angka per sampel, jadi tidak bisa dijadikan gambar.

Pixel membawa sinyal kuat, cukup untuk membedakan digit dengan baik. Morfologi membawa sinyal lemah, yaitu 6 angka kasar yang banyak tumpang tindih antar digit. Perbedaan kekuatan sinyal inilah yang nanti memunculkan masalah modalitas terabaikan.

```python
# Bentuk dan isi mentah kedua modalitas.
digit = pixel.target.astype(int) - 1   # label mfeat '1'..'10' -> digit 0..9

print(f"Jumlah sampel total    : {Xa_raw.shape[0]}")
print(f"Modalitas A (Pixel)    : {Xa_raw.shape}, nilai {int(Xa_raw.min())}-{int(Xa_raw.max())}")
print(f"Modalitas B (Morfologi): {Xb_raw.shape}")
uniq, cnt = np.unique(digit, return_counts=True)
print("Sampel per digit       :", dict(zip(uniq.tolist(), cnt.tolist())))
print()
print("Satu sampel Pixel (5 dari 240 angka):", Xa_raw[0, :5])
print("Satu sampel Morfologi (6 angka)     :", np.round(Xb_raw[0], 2))
```

## 2. Late Fusion Model + Baseline


```python
class LateFusionModel(nn.Module):
    def __init__(self, d_a=240, d_b=6, d_hidden=32, num_classes=10, modality_dropout=0.0):
        super().__init__()
        self.enc_a = nn.Sequential(
            nn.Linear(d_a, d_hidden), nn.ReLU(), nn.Dropout(modality_dropout))
        self.enc_b = nn.Sequential(
            nn.Linear(d_b, d_hidden), nn.ReLU(), nn.Dropout(modality_dropout))
        self.head = nn.Linear(d_hidden * 2, num_classes)

    def forward(self, xa, xb):
        ha = self.enc_a(xa)
        hb = self.enc_b(xb)
        fused = torch.cat([ha, hb], dim=1)
        return self.head(fused)

def evaluate(model, dl_va, shuffle_a=False, shuffle_b=False, zero_a=False, zero_b=False):
    """Akurasi validasi. Opsi mengubah satu modalitas untuk ablation:
    zero_* mengosongkan modalitas, shuffle_* mengacak modalitas antar-sampel."""
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for Xa, Xb, yb in dl_va:
            if zero_a:    Xa = torch.zeros_like(Xa)
            if zero_b:    Xb = torch.zeros_like(Xb)
            if shuffle_a: Xa = Xa[torch.randperm(len(Xa))]
            if shuffle_b: Xb = Xb[torch.randperm(len(Xb))]
            Xa, Xb, yb = Xa.to(DEVICE), Xb.to(DEVICE), yb.to(DEVICE)
            pred = model(Xa, Xb).argmax(1)
            correct += (pred == yb).sum().item()
            total += len(yb)
    return correct / total

def train_model(model, dl_tr, dl_va, epochs=30, lr=5e-4):
    opt = optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    crit = nn.CrossEntropyLoss()
    history = {"train_loss": [], "val_acc": []}
    for epoch in range(epochs):
        model.train()
        running = 0.0
        for Xa, Xb, yb in dl_tr:
            Xa, Xb, yb = Xa.to(DEVICE), Xb.to(DEVICE), yb.to(DEVICE)
            opt.zero_grad()
            loss = crit(model(Xa, Xb), yb)
            loss.backward()
            opt.step()
            running += loss.item() * len(Xa)
        history["train_loss"].append(running / len(dl_tr.dataset))
        history["val_acc"].append(evaluate(model, dl_va))
    return history
```

```python
model = LateFusionModel(d_a=D_A, d_b=D_B).to(DEVICE)
hist = train_model(model, dl_tr, dl_va)

plt.figure(figsize=(10, 3))
plt.subplot(121)
plt.plot(hist["train_loss"]); plt.title("Train Loss (CrossEntropy)"); plt.xlabel("epoch"); plt.grid(alpha=0.3)
plt.subplot(122)
plt.plot(hist["val_acc"]); plt.title("Val Accuracy"); plt.xlabel("epoch"); plt.grid(alpha=0.3)
plt.tight_layout(); plt.show()

acc_fusion = evaluate(model, dl_va)
print(f"Late Fusion (kedua modalitas) - Val Accuracy: {acc_fusion:.4f}")
```

## 3. Ablation Per-Modalitas (6 kondisi)

Kita melatih satu model fusion, lalu menguji enam kondisi pada validation set:

1. **Fusion (keduanya)** - kedua modalitas utuh, sebagai baseline.
2. **Pixel only** - modalitas Morfologi dikosongkan.
3. **Morfologi only** - modalitas Pixel dikosongkan.
4. **Random Pixel** - modalitas Pixel diacak antar-sampel saat eval.
5. **Random Morfologi** - modalitas Morfologi diacak antar-sampel saat eval.
6. **Random both** - kedua modalitas diacak.

Bandingkan: jika akurasi turun sedikit saat satu modalitas diacak, modalitas itu memang diabaikan (modalitas terabaikan).


```python
# Pakai model fusion yang sudah dilatih di atas. Semua kondisi diukur pada model yang sama.
print("=== HASIL ABLATION (akurasi, makin tinggi makin baik) ===\n")
results = {}
results["Fusion (both)"]    = acc_fusion
results["Pixel only"]       = evaluate(model, dl_va, zero_b=True)
results["Morfologi only"]   = evaluate(model, dl_va, zero_a=True)
results["Random Pixel"]     = evaluate(model, dl_va, shuffle_a=True)
results["Random Morfologi"] = evaluate(model, dl_va, shuffle_b=True)
results["Random both"]      = evaluate(model, dl_va, shuffle_a=True, shuffle_b=True)

print(f"{'Kondisi':20s} {'Val Acc':>8s} {'vs Fusion':>10s}")
print("-" * 42)
for cond, acc in sorted(results.items(), key=lambda x: -x[1]):
    print(f"{cond:20s} {acc:8.4f} {acc - acc_fusion:+10.4f}")
```

## 4. Diagnostik Visual


```python
# Visualisasi enam kondisi
conds = list(results.keys())
vals = [results[c] for c in conds]
colors = ["#2ecc71", "#3498db", "#e67e22", "#9b59b6", "#95a5a6", "#e74c3c"]

plt.figure(figsize=(10, 5))
bars = plt.bar(range(len(vals)), vals, color=colors[:len(vals)])
plt.xticks(range(len(vals)), conds, rotation=15, ha="right")
plt.ylabel("Val Accuracy")
plt.title("Ablation Multimodal: Kontribusi Per-Modalitas")
plt.axhline(y=0.1, color="gray", linestyle="--", label="Tebak acak (0.10)")
plt.axhline(y=results["Random both"], color="brown", linestyle=":",
            label=f"Random both ({results['Random both']:.3f})")
plt.legend(); plt.grid(alpha=0.3, axis="y")
for bar, v in zip(bars, vals):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
             f"{v:.3f}", ha="center", va="bottom", fontsize=9)
plt.tight_layout(); plt.show()

# Diagnosis adaptif: seberapa besar akurasi turun saat tiap modalitas diacak
drop_a = acc_fusion - results["Random Pixel"]      # kontribusi Pixel
drop_b = acc_fusion - results["Random Morfologi"]  # kontribusi Morfologi
print("\n=== DIAGNOSIS ===")
print(f"Penurunan akurasi saat Pixel diacak    : {drop_a:+.4f}")
print(f"Penurunan akurasi saat Morfologi diacak: {drop_b:+.4f}")
print()
TOL = 0.05
if drop_b < TOL and drop_a >= TOL:
    print(">>> Modalitas MORFOLOGI diabaikan: mengacaknya hampir tidak menurunkan akurasi.")
elif drop_a < TOL and drop_b >= TOL:
    print(">>> Modalitas PIXEL diabaikan: mengacaknya hampir tidak menurunkan akurasi.")
elif drop_a < TOL and drop_b < TOL:
    print(">>> Dua-duanya nyaris tidak berpengaruh: periksa apakah model benar-benar belajar.")
else:
    print(">>> Kedua modalitas berkontribusi (mengacak salah satu menurunkan akurasi).")
    print("    Tetap perlu modality dropout agar model robust saat satu modalitas hilang.")
```

## 5. Modality Dropout

Training dengan mengosongkan salah satu modalitas secara acak (peluang 15%). Ini memaksa model belajar representasi yang berguna dari tiap modalitas secara mandiri.


```python
# Modality dropout: selama training, kadang-kadang kosongkan salah satu modalitas.
# Ini memaksa model untuk tidak bergantung penuh pada satu modalitas saja.

model_md = LateFusionModel(d_a=D_A, d_b=D_B, modality_dropout=0.2).to(DEVICE)
opt = optim.AdamW(model_md.parameters(), lr=5e-4, weight_decay=1e-4)
crit = nn.CrossEntropyLoss()
history_md = []
for epoch in range(40):
    model_md.train()
    running = 0.0
    for Xa, Xb, yb in dl_tr:
        Xa, Xb, yb = Xa.to(DEVICE), Xb.to(DEVICE), yb.to(DEVICE)
        # peluang 15% mengosongkan Pixel, peluang 15% mengosongkan Morfologi
        if torch.rand(1).item() < 0.15:
            Xa = torch.zeros_like(Xa)
        if torch.rand(1).item() < 0.15:
            Xb = torch.zeros_like(Xb)
        opt.zero_grad()
        loss = crit(model_md(Xa, Xb), yb)
        loss.backward()
        opt.step()
        running += loss.item() * len(Xa)
    history_md.append(running / len(dl_tr.dataset))

# Evaluasi dengan tiap modalitas sendirian (kosongkan modalitas yang lain)
md_pixel = evaluate(model_md, dl_va, zero_b=True)   # Pixel only
md_morph = evaluate(model_md, dl_va, zero_a=True)   # Morfologi only

print("=== Hasil Modality Dropout ===")
print(f"Pixel only (model dilatih dengan MD)    : {md_pixel:.4f}")
print(f"Morfologi only (model dilatih dengan MD): {md_morph:.4f}")
print()
print("Bandingkan dengan model tanpa MD:")
print(f"Pixel only:     {results['Pixel only']:.4f} -> {md_pixel:.4f}")
print(f"Morfologi only: {results['Morfologi only']:.4f} -> {md_morph:.4f}")
print()
if md_pixel > results['Pixel only'] and md_morph > results['Morfologi only']:
    print(">>> Modality dropout menaikkan akurasi tiap modalitas tunggal. Berhasil!")
elif md_morph > results['Morfologi only']:
    print(">>> Modalitas Morfologi yang tadinya diabaikan kini jauh lebih berguna sendirian.")
elif md_pixel > results['Pixel only']:
    print(">>> Modalitas Pixel jadi lebih robust.")
else:
    print(">>> Modality dropout belum menaikkan performa modalitas tunggal; coba naikkan peluang dropout.")

plt.figure(figsize=(8, 3))
plt.plot(history_md); plt.title("Loss Training (Modality Dropout)")
plt.xlabel("epoch"); plt.grid(alpha=0.3); plt.show()
```
